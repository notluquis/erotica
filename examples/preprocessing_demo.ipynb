````python
# Configuración del entorno
import os
import sys
from pathlib import Path
import warnings
warnings.filterwarnings('ignore')

# Configurar rutas del proyecto
PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'examples' else Path.cwd()
DATA_ROOT = PROJECT_ROOT / 'data'

print(f"📁 PROJECT_ROOT: {PROJECT_ROOT}")
print(f"📁 DATA_ROOT: {DATA_ROOT}")

# Verificar que estamos en un entorno COSMIC
try:
    import cosmic
    print(f"✅ COSMIC disponible")
except ImportError:
    print("❌ COSMIC no encontrado, instalando en modo desarrollo...")
    os.system(f"pip install -e {PROJECT_ROOT}")
    import cosmic
    print("✅ COSMIC instalado e importado")
````

````python
# Imports usando la nueva estructura modular
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Nueva estructura modular (recomendado)
from cosmic.io.loader import DataLoader
from cosmic.preprocess.preprocessor import DataPreprocessor
from cosmic.core.clustering import Clustering
from cosmic.analysis.analyzer import ClusterAnalyzer

# También funcionan los imports legacy para compatibilidad
# from data_loader import DataLoader
# from data_preprocessor import DataPreprocessor
# from clustering import HDBSCANClustering as Clustering

print("📦 Todos los módulos importados correctamente")
print(f"🔧 DataLoader: {DataLoader}")
print(f"🔧 DataPreprocessor: {DataPreprocessor}")
print(f"🔧 Clustering: {Clustering}")
print(f"🔧 ClusterAnalyzer: {ClusterAnalyzer}")
````

## 1. Carga de Datos

Usando el nuevo sistema modular de carga de datos.

````python
# Configurar el cargador de datos
data_file = DATA_ROOT / "data.ecsv"  # Datos principales
print(f"📄 Archivo de datos: {data_file}")
print(f"📊 Existe: {data_file.exists()}")

# Inicializar el cargador con la nueva API
loader = DataLoader(
    file_path=str(data_file),
    verbose=True
)

# Cargar los datos
try:
    data = loader.load_data()
    print(f"✅ Datos cargados: {len(data)} filas, {len(data.columns)} columnas")
    print(f"🏷️  Columnas disponibles: {list(data.columns)}")
except Exception as e:
    print(f"❌ Error cargando datos: {e}")
    # Usar datos de ejemplo si hay problemas
    print("🔄 Generando datos de ejemplo...")
    
    # Datos sintéticos para demostración
    np.random.seed(42)
    n_stars = 1000
    
    data = pd.DataFrame({
        'ra': np.random.uniform(250, 260, n_stars),
        'dec': np.random.uniform(-30, -20, n_stars), 
        'pmra': np.random.normal(-5, 2, n_stars),
        'pmdec': np.random.normal(3, 2, n_stars),
        'Gmag': np.random.uniform(12, 18, n_stars),
        'parallax': np.random.uniform(1, 3, n_stars),
        'source_id': range(n_stars)
    })
    print(f"✅ Datos sintéticos generados: {len(data)} estrellas")
````

## 2. Preprocesamiento

Usando el nuevo sistema modular de preprocesamiento.

````python
# Inicializar el preprocesador con la nueva API
preprocessor = DataPreprocessor(
    parallax_snr_limit=5.0,  # Límite S/N para parallax
    pm_snr_limit=5.0,        # Límite S/N para proper motions
    magnitude_limits=(12, 18), # Límites de magnitud
    verbose=True
)

print("🔧 Preprocesador configurado")
print(f"📊 Datos originales: {len(data)} filas")

# Aplicar preprocesamiento
try:
    processed_data = preprocessor.preprocess(data)
    print(f"✅ Preprocesamiento completado")
    print(f"📊 Datos procesados: {len(processed_data)} filas")
    print(f"📉 Filtrados: {len(data) - len(processed_data)} filas ({100*(len(data) - len(processed_data))/len(data):.1f}%)")
    
    # Mostrar estadísticas
    if len(processed_data) > 0:
        print(f"\n📈 Estadísticas de datos procesados:")
        for col in ['pmra', 'pmdec', 'Gmag', 'parallax']:
            if col in processed_data.columns:
                mean_val = processed_data[col].mean()
                std_val = processed_data[col].std()
                print(f"   {col}: {mean_val:.3f} ± {std_val:.3f}")
                
except Exception as e:
    print(f"❌ Error en preprocesamiento: {e}")
    processed_data = data.copy()
    print("🔄 Usando datos sin procesar")
````

## 3. Clustering

Usando el nuevo sistema modular de clustering.

````python
# Configurar el clustering con la nueva API
clustering = Clustering(
    feature_columns=['pmra', 'pmdec'],  # Columnas para clustering
    min_cluster_size=10,                # Tamaño mínimo de cluster
    min_samples=5,                      # Muestras mínimas
    verbose=True
)

print("🔧 Clustering configurado")
print(f"📊 Características: {clustering.feature_columns}")

# Ejecutar clustering
try:
    results = clustering.fit_predict(processed_data)
    print(f"✅ Clustering completado")
    
    # Estadísticas de clustering
    unique_clusters = np.unique(results['cluster'])
    n_clusters = len(unique_clusters[unique_clusters >= 0])  # Excluir ruido (-1)
    n_noise = np.sum(results['cluster'] == -1)
    
    print(f"🔍 Clusters encontrados: {n_clusters}")
    print(f"📊 Estrellas en clusters: {len(results) - n_noise}")
    print(f"🌫️  Ruido: {n_noise} estrellas ({100*n_noise/len(results):.1f}%)")
    
    # Mostrar tamaños de clusters
    if n_clusters > 0:
        cluster_sizes = pd.Series(results['cluster']).value_counts().sort_index()
        cluster_sizes = cluster_sizes[cluster_sizes.index >= 0]  # Excluir ruido
        print(f"\n📏 Tamaños de clusters:")
        for cluster_id, size in cluster_sizes.head(10).items():
            print(f"   Cluster {cluster_id}: {size} estrellas")
        if len(cluster_sizes) > 10:
            print(f"   ... y {len(cluster_sizes) - 10} clusters más")
            
except Exception as e:
    print(f"❌ Error en clustering: {e}")
    # Crear resultados dummy
    results = processed_data.copy()
    results['cluster'] = -1  # Todo como ruido
    n_clusters = 0
````

## 4. Análisis de Resultados

Usando el nuevo sistema modular de análisis.

````python
# Solo continuar si tenemos clusters
if n_clusters > 0:
    try:
        # Crear analizador con la nueva API
        analyzer = ClusterAnalyzer(
            data=results,
            output_dir=DATA_ROOT / "analysis_output"
        )
        
        print("🔧 Analizador configurado")
        
        # Generar resumen de clusters
        summary = analyzer.clusters_summary()
        print(f"✅ Resumen generado para {len(summary)} clusters")
        print("\n📊 Primeros clusters:")
        print(summary.head())
        
        # Crear visualizaciones básicas
        print("\n🎨 Generando visualizaciones...")
        
        # Plot de proper motions
        fig, ax = plt.subplots(figsize=(10, 8))
        
        # Plotear ruido
        noise_mask = results['cluster'] == -1
        if noise_mask.sum() > 0:
            ax.scatter(results.loc[noise_mask, 'pmra'], 
                      results.loc[noise_mask, 'pmdec'],
                      c='lightgray', s=10, alpha=0.5, label='Ruido')
        
        # Plotear clusters
        colors = plt.cm.tab10(np.linspace(0, 1, min(n_clusters, 10)))
        for i, cluster_id in enumerate(unique_clusters[unique_clusters >= 0][:10]):
            cluster_mask = results['cluster'] == cluster_id
            cluster_data = results[cluster_mask]
            
            ax.scatter(cluster_data['pmra'], cluster_data['pmdec'],
                      c=[colors[i]], s=30, alpha=0.7, 
                      label=f'Cluster {cluster_id} (n={len(cluster_data)})')
        
        ax.set_xlabel('Proper Motion RA (mas/yr)')
        ax.set_ylabel('Proper Motion Dec (mas/yr)')
        ax.set_title(f'COSMIC Clustering Results - {n_clusters} Clusters Found')
        ax.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
        ax.grid(True, alpha=0.3)
        
        plt.tight_layout()
        plt.show()
        
        print(f"✅ Análisis completado exitosamente")
        
    except Exception as e:
        print(f"❌ Error en análisis: {e}")
        print("📊 Datos disponibles para análisis manual")
        
else:
    print("⚠️  No se encontraron clusters para analizar")
    print("💡 Sugerencias:")
    print("   - Reducir min_cluster_size")
    print("   - Verificar calidad de los datos")
    print("   - Ajustar parámetros de preprocesamiento")
````
